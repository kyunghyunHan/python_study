```python
#comprehension
#list set dict를 만드는 기법
a =[10,20,30]
[x for x in a]
{x for x in a}
b = {x: x * 2 for x in a}
print(b)
```

```python
# expression(표현식)
# - 평가(evaluate)하면 하나의 값(object)이 나오는 코드
#
# 예:
# 10, 3.14, "python", [1, 2], (1,)
# a + b
# 함수 호출
# conditional expression
# comprehension
#
# expression은 다음과 같은 곳에서 사용 가능
# - 이름에 대입: a = 표현식
# - 함수 argument: print(표현식)
# - return: return 표현식
# - 괄호: (표현식)
# - if/while 조건: if 표현식:
#
# Python에서는 정수도 object
a = 10
print(a.bit_length())  # a가 가리키는 int object의 메서드 호출
```

```python
[x*2 for x in range(1,5) if not x %2] #파이썬에서 0은 false
a = [[x ,y] for x ,y in [(1,2),(2,3),(4,5)]]
print(a)
```
