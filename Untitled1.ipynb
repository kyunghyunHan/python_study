```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path


# =====================================================
# 0. 기본 설정
# =====================================================

FILE = "motor_current_features.csv"

# 그래프 저장 폴더
OUT = Path("analysis_results")
OUT.mkdir(exist_ok=True)

# macOS 한글 폰트
plt.rcParams["font.family"] = "AppleGothic"
plt.rcParams["axes.unicode_minus"] = False


# =====================================================
# 1. CSV 불러오기
# =====================================================

df = pd.read_csv(FILE)

print("===================================")
print("1. 전체 데이터 정보")
print("===================================")

print("데이터 개수:", len(df))
print("컬럼 개수:", len(df.columns))

print()
print("처음 5행")
print(df.head())

print()
print("데이터 정보")
df.info()


# =====================================================
# 2. 시간 변환
# =====================================================

df["recorded_at"] = pd.to_datetime(
    df["recorded_at"],
    errors="coerce"
)


# =====================================================
# 3. 데이터 품질 검사
# =====================================================

print()
print("===================================")
print("2. 데이터 품질 검사")
print("===================================")

print()
print("전처리 상태")
print(df["status"].value_counts(dropna=False))

print()
print("라벨 불일치")
print(df["label_mismatch"].value_counts(dropna=False))

print()
print("센서 종류")
print(df["sensor"].value_counts(dropna=False))

print()
print("Training / Validation")
print(df["split"].value_counts(dropna=False))

print()
print("결측치 개수")
print(
    df.isnull()
      .sum()
      .sort_values(ascending=False)
      .head(20)
)


# =====================================================
# 4. 정상적으로 처리된 데이터만 선택
# =====================================================

clean = df[
    (df["status"] == "ok")
    & (df["sensor"] == "current")
    & (df["label"].notna())
    & (df["label_mismatch"].astype(str).str.lower() == "false")
].copy()

print()
print("정상 처리된 데이터:", len(clean))


# =====================================================
# 5. 고장 유형별 개수
# =====================================================

print()
print("===================================")
print("3. 고장 유형별 개수")
print("===================================")

label_count = clean["label"].value_counts()

print(label_count)

plt.figure(figsize=(12, 5))

label_count.plot(
    kind="bar"
)

plt.title("고장 유형별 데이터 개수")
plt.xlabel("고장 유형")
plt.ylabel("데이터 개수")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()

plt.savefig(OUT / "01_label_count.png", dpi=150)
plt.show()


# =====================================================
# 6. 모터 출력별 데이터 분포
# =====================================================

print()
print("===================================")
print("4. 모터 출력별 데이터 분포")
print("===================================")

power_label = pd.crosstab(
    clean["power_kw"],
    clean["label"]
)

print(power_label)

power_label.plot(
    kind="bar",
    figsize=(12, 5)
)

plt.title("모터 출력별 고장 유형 분포")
plt.xlabel("모터 출력 (kW)")
plt.ylabel("데이터 개수")
plt.xticks(rotation=0)
plt.legend(title="고장 유형")
plt.tight_layout()

plt.savefig(OUT / "02_power_label.png", dpi=150)
plt.show()


# =====================================================
# 7. 모터 모델별 고장 유형 분포
# =====================================================

print()
print("===================================")
print("5. 모터 모델별 고장 분포")
print("===================================")

model_label = pd.crosstab(
    clean["motor_model"],
    clean["label"]
)

print(model_label)

model_label.plot(
    kind="bar",
    stacked=True,
    figsize=(13, 6)
)

plt.title("모터 모델별 정상 / 불량 데이터")
plt.xlabel("모터 모델")
plt.ylabel("데이터 개수")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()

plt.savefig(OUT / "03_model_label.png", dpi=150)
plt.show()


# =====================================================
# 8. 고장 유형별 RMS 통계
#
# RMS = 전류 실효값
# =====================================================

print()
print("===================================")
print("6. 고장 유형별 RMS")
print("===================================")

rms_columns = [
    "ch1_rms",
    "ch2_rms",
    "ch3_rms"
]

print(
    clean.groupby("label")[rms_columns]
         .agg(["mean", "median", "std"])
         .round(4)
)


# =====================================================
# 9. 2.2kW 모터만 선택
#
# 출력이 다른 모터는 전류 크기가 다르므로
# 우선 같은 출력끼리 비교
# =====================================================

motor = clean[
    clean["power_kw"] == 2.2
].copy()

print()
print("2.2kW 데이터:", len(motor))


# =====================================================
# 10. 정상/불량 RMS Boxplot
# =====================================================

labels = motor["label"].dropna().unique()

box_data = [
    motor.loc[
        motor["label"] == label,
        "ch1_rms"
    ].dropna().values
    for label in labels
]

if len(box_data) > 0:

    plt.figure(figsize=(12, 5))

    plt.boxplot(
        box_data,
        tick_labels=labels,
        showfliers=True
    )

    plt.title("고장 유형별 CH1 전류 RMS 분포 (2.2kW)")
    plt.xlabel("고장 유형")
    plt.ylabel("CH1 RMS")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()

    plt.savefig(OUT / "04_rms_boxplot.png", dpi=150)
    plt.show()


# =====================================================
# 11. CH1 / CH2 / CH3 RMS 비교
# =====================================================

rms_mean = motor.groupby("label")[
    [
        "ch1_rms",
        "ch2_rms",
        "ch3_rms"
    ]
].mean()

print()
print("채널별 RMS 평균")
print(rms_mean.round(4))

if len(rms_mean) > 0:

    rms_mean.plot(
        kind="bar",
        figsize=(12, 5)
    )

    plt.title("고장 유형별 3채널 전류 RMS 평균")
    plt.xlabel("고장 유형")
    plt.ylabel("평균 RMS")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()

    plt.savefig(OUT / "05_channel_rms.png", dpi=150)
    plt.show()


# =====================================================
# 12. FFT 특징 분석
#
# 어떤 주파수 대역에서 에너지가 큰지 확인
# =====================================================

fft_features = [
    "ch1_fft_energy_0_10_hz",
    "ch1_fft_energy_10_50_hz",
    "ch1_fft_energy_50_100_hz",
    "ch1_fft_energy_100_200_hz",
    "ch1_fft_energy_200_500_hz",
    "ch1_fft_energy_500_1000_hz"
]

fft_mean = motor.groupby("label")[
    fft_features
].mean()

print()
print("===================================")
print("7. FFT 에너지")
print("===================================")

print(fft_mean.round(4))

# 에너지 크기 차이가 클 수 있어 log1p 적용
# log1p(x) = log(1+x)

if len(fft_mean) > 0:

    fft_plot = np.log1p(fft_mean.clip(lower=0))

    fft_plot.T.plot(
        kind="bar",
        figsize=(13, 6)
    )

    plt.title("고장 유형별 FFT 주파수 에너지")
    plt.xlabel("주파수 대역")
    plt.ylabel("log(1 + 평균 에너지)")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()

    plt.savefig(OUT / "06_fft_energy.png", dpi=150)
    plt.show()


# =====================================================
# 13. 가장 강한 FFT 주파수 분석
# =====================================================

print()
print("===================================")
print("8. 주요 주파수")
print("===================================")

print(
    motor.groupby("label")[
        [
            "ch1_fft_peak1_hz",
            "ch2_fft_peak1_hz",
            "ch3_fft_peak1_hz"
        ]
    ].median()
)


# =====================================================
# 14. RMS가 매우 낮은 데이터 확인
#
# 0.1은 임시 분석 기준
# 실제 고장 기준은 아님
# =====================================================

print()
print("===================================")
print("9. 저전류 데이터 검사")
print("===================================")

LOW_CURRENT = 0.1

low_current = clean[
    clean["ch1_rms"] < LOW_CURRENT
].copy()

print("저전류 데이터:", len(low_current))

print(
    low_current[
        [
            "motor_model",
            "label",
            "recorded_at",
            "ch1_rms",
            "ch2_rms",
            "ch3_rms"
        ]
    ].head(30).to_string(index=False)
)

low_current.to_csv(
    OUT / "low_current.csv",
    index=False,
    encoding="utf-8-sig"
)


# =====================================================
# 15. 특징 간 상관관계
# =====================================================

features = [
    "ch1_mean",
    "ch1_std",
    "ch1_rms",
    "ch1_peak_to_peak",
    "ch1_crest_factor",
    "ch1_skewness",
    "ch1_kurtosis",
    "ch1_fft_peak1_hz",
    "ch1_fft_ac_energy"
]

corr = motor[features].corr()

print()
print("===================================")
print("10. 특징 상관관계")
print("===================================")

print(corr.round(3))

plt.figure(figsize=(11, 8))

plt.imshow(
    corr,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.colorbar(label="상관계수")

plt.xticks(
    range(len(features)),
    features,
    rotation=90
)

plt.yticks(
    range(len(features)),
    features
)

plt.title("전류 특징 상관관계 히트맵")
plt.tight_layout()

plt.savefig(OUT / "07_correlation.png", dpi=150)
plt.show()


# =====================================================
# 16. 시간에 따른 전류 RMS 변화
#
# 모터 모델 하나 + 장비 ID 하나 선택
# =====================================================

print()
print("===================================")
print("11. 시간별 전류 변화")
print("===================================")

if len(motor) > 0:

    # 가장 데이터가 많은 장비 선택
    device = motor["device_id"].value_counts().idxmax()

    time_data = motor[
        motor["device_id"] == device
    ].copy()

    time_data = time_data.dropna(
        subset=["recorded_at"]
    )

    time_data = time_data.sort_values(
        "recorded_at"
    )

    plt.figure(figsize=(13, 5))

    for column in [
        "ch1_rms",
        "ch2_rms",
        "ch3_rms"
    ]:
        plt.plot(
            time_data["recorded_at"],
            time_data[column],
            label=column,
            marker=".",
            markersize=3
        )

    plt.title(f"시간에 따른 전류 RMS 변화 - {device}")
    plt.xlabel("측정 시간")
    plt.ylabel("RMS")
    plt.legend()
    plt.xticks(rotation=30)
    plt.tight_layout()

    plt.savefig(OUT / "08_time_rms.png", dpi=150)
    plt.show()


# =====================================================
# 17. 분석 결과 저장
# =====================================================

clean.to_csv(
    OUT / "clean_current_features.csv",
    index=False,
    encoding="utf-8-sig"
)

power_label.to_csv(
    OUT / "power_label_count.csv",
    encoding="utf-8-sig"
)

model_label.to_csv(
    OUT / "motor_model_label_count.csv",
    encoding="utf-8-sig"
)

corr.to_csv(
    OUT / "feature_correlation.csv",
    encoding="utf-8-sig"
)


print()
print("===================================")
print("분석 완료")
print("===================================")

print("저장 위치:", OUT.resolve())
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import joblib

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

from sklearn.ensemble import (
    RandomForestClassifier,
    GradientBoostingClassifier
)

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    ConfusionMatrixDisplay,
    f1_score
)


# =====================================================
# 0. 기본 설정
# =====================================================

plt.rcParams["font.family"] = "AppleGothic"
plt.rcParams["axes.unicode_minus"] = False

RANDOM_STATE = 1234


# =====================================================
# 1. 데이터 불러오기
# =====================================================

df = pd.read_csv("motor_current_features.csv")

# 정상적으로 처리된 전류 데이터만 선택
df = df[
    (df["status"] == "ok") &
    (df["sensor"] == "current") &
    (df["label"].notna()) &
    (
        df["label_mismatch"]
        .astype(str)
        .str.lower() == "false"
    )
].copy()

# 2.2kW 모터만 선택
df = df[df["power_kw"] == 2.2].copy()

print("전체 데이터:", len(df))

print("\n고장 유형:")
print(df["label"].value_counts())


# =====================================================
# 2. X (특징) 선택
# =====================================================

feature_suffixes = [
    "mean",
    "std",
    "min",
    "max",
    "rms",
    "peak_to_peak",
    "crest_factor",
    "skewness",
    "kurtosis",
    "fft_peak1_hz",
    "fft_peak2_hz",
    "fft_peak3_hz",
    "fft_ac_energy",
    "fft_energy_0_10_hz",
    "fft_energy_10_50_hz",
    "fft_energy_50_100_hz",
    "fft_energy_100_200_hz",
    "fft_energy_200_500_hz",
    "fft_energy_500_1000_hz"
]

features = [
    f"ch{ch}_{suffix}"
    for ch in [1, 2, 3]
    for suffix in feature_suffixes
]

# 실제 CSV에 존재하는 컬럼만 사용
features = [
    col for col in features
    if col in df.columns
]

X = df[features].apply(
    pd.to_numeric,
    errors="coerce"
)

X = X.replace(
    [np.inf, -np.inf],
    np.nan
)

# Y = 실제 정상/고장 유형
Y = df["label"]

print("\n사용 특징 개수:", len(features))


# =====================================================
# 3. Train / Test 분리
#
# Train = 80%
# Test = 20%
#
# stratify=Y
# -> 고장 유형의 비율을 비슷하게 유지
# =====================================================

x_train, x_test, y_train, y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    stratify=Y,
    random_state=RANDOM_STATE
)

print("\n================================")
print("Train / Test 데이터 분리")
print("================================")

print("Train:", len(x_train))
print("Test :", len(x_test))

print("\nTrain 고장 유형:")
print(y_train.value_counts())

print("\nTest 고장 유형:")
print(y_test.value_counts())


# =====================================================
# 4. 머신러닝 모델 정의
# =====================================================

models = {
    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        max_depth=10,
        min_samples_leaf=3,
        random_state=RANDOM_STATE,
        n_jobs=-1
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=100,
        max_depth=3,
        random_state=RANDOM_STATE
    )
}


# =====================================================
# 5. 모델 학습 및 예측
# =====================================================

results = []
trained_models = {}

for name, classifier in models.items():

    # 결측값을 Train의 중앙값으로 채움
    model = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("classifier", classifier)
    ])

    # ★ 머신러닝 학습
    model.fit(x_train, y_train)

    # ★ 테스트 데이터 예측
    y_pred = model.predict(x_test)

    # Train 정확도
    train_score = model.score(
        x_train,
        y_train
    )

    # Test 정확도
    test_score = accuracy_score(
        y_test,
        y_pred
    )

    # 다중 분류에서도 사용할 수 있는 F1
    macro_f1 = f1_score(
        y_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    results.append({
        "모델": name,
        "Train 정확도": train_score,
        "Test 정확도": test_score,
        "Macro F1": macro_f1
    })

    trained_models[name] = model

    print("\n================================")
    print(name)
    print("================================")

    print("Train 정확도:", round(train_score, 4))
    print("Test 정확도 :", round(test_score, 4))
    print("Macro F1    :", round(macro_f1, 4))

    print("\nClassification Report")

    print(
        classification_report(
            y_test,
            y_pred,
            zero_division=0
        )
    )

    # =================================================
    # 6. Confusion Matrix 그래프
    # =================================================

    ConfusionMatrixDisplay.from_predictions(
        y_test,
        y_pred,
        xticks_rotation=45,
        cmap="Blues"
    )

    plt.title(f"{name} - Confusion Matrix")
    plt.tight_layout()
    plt.show()


# =====================================================
# 7. 모델 성능 비교
# =====================================================

result_df = pd.DataFrame(results)

print("\n================================")
print("모델 성능 비교")
print("================================")

print(result_df.round(4).to_string(index=False))

result_df.set_index("모델")[
    [
        "Train 정확도",
        "Test 정확도",
        "Macro F1"
    ]
].plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("모델별 성능 비교")
plt.ylabel("Score")
plt.ylim(0, 1.05)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# =====================================================
# 8. 가장 높은 Macro F1 모델 선택
# =====================================================

best_name = result_df.loc[
    result_df["Macro F1"].idxmax(),
    "모델"
]

best_model = trained_models[best_name]

print("\n선택된 모델:", best_name)


# =====================================================
# 9. Feature Importance
# =====================================================

classifier = best_model.named_steps["classifier"]

importance = pd.Series(
    classifier.feature_importances_,
    index=features
).sort_values(ascending=False)

print("\n================================")
print("중요 특징 TOP 15")
print("================================")

print(importance.head(15))

importance.head(15).sort_values().plot(
    kind="barh",
    figsize=(11, 7)
)

plt.title(f"{best_name} - Feature Importance")
plt.xlabel("Importance")
plt.tight_layout()
plt.show()


# =====================================================
# 10. 실제값 / 예측값 비교
# =====================================================

y_pred = best_model.predict(x_test)

comparison = df.loc[x_test.index, [
    "source_file",
    "motor_model",
    "label"
]].copy()

comparison["예측"] = y_pred

comparison["정답여부"] = (
    comparison["label"] == comparison["예측"]
)

print("\n================================")
print("실제값 / 예측값 비교")
print("================================")

print(comparison.head(30).to_string(index=False))


# =====================================================
# 11. 모델 저장
# =====================================================

joblib.dump(
    best_model,
    "motor_current_model.pkl"
)

print("\n모델 저장 완료: motor_current_model.pkl")
```

```python
import matplotlib.pyplot as plt

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    ConfusionMatrixDisplay
)

# ======================================
# 1. SVM 모델 생성
# ======================================

svm_model = Pipeline([
    # 결측치 처리
    ("imputer", SimpleImputer(strategy="median")),

    # SVM은 특징의 크기에 영향을 받으므로 표준화
    ("scaler", StandardScaler()),

    # SVM 분류 모델
    ("svm", SVC(
        kernel="rbf",
        C=1.0,
        gamma="scale",
        random_state=1234
    ))
])


# ======================================
# 2. 머신러닝 학습
# ======================================

svm_model.fit(x_train, y_train)


# ======================================
# 3. 예측
# ======================================

y_pred = svm_model.predict(x_test)


# ======================================
# 4. 성능 평가
# ======================================

print("===== SVM 결과 =====")

print(
    "Train 정확도:",
    round(svm_model.score(x_train, y_train), 4)
)

print(
    "Test 정확도:",
    round(accuracy_score(y_test, y_pred), 4)
)

print(
    "Macro F1:",
    round(f1_score(
        y_test,
        y_pred,
        average="macro"
    ), 4)
)

print()
print(classification_report(y_test, y_pred))


# ======================================
# 5. Confusion Matrix
# ======================================

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    cmap="Blues",
    xticks_rotation=45
)

plt.title("SVM Confusion Matrix")
plt.tight_layout()
plt.show()
```
