```python
#pack 여러개를 하나의 튜플로 묶음
a  =1,2,3
#unpack 컨테이너에있는 여러 데이터를 각가의 이름에 나누는것
x, y ,z = a 
x, y , z = 4,5,6 #언팩 + 팩
print(x,y,z)
for x , y in [(1,2),(3,4)]:
    print(x, y)
```
