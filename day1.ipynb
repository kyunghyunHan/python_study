```python
print(round(2.5),round(3.5),round(-2.5))
#가까운 짝수로 반올림 .5일때만 .51은 아님 

def my_round(x):
    lower = int(x)
    diff = x - lower

    if diff < 0.5:
        return lower

    if diff > 0.5:
        return lower + 1

    # 정확히 .5인 경우
    if lower % 2 == 0:
        return lower
    else:
        return lower + 1

print(my_round(2.5),my_round(3.5),my_round(-2.5))
```
