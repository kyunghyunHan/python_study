# 머신러닝

## 선형회귀 (Linear Regression)
- 숫자 예측
- 연속값 예측
- 예: 집값, 온도, 매출

## 로지스틱 회귀 (Logistic Regression)
- 확률을 이용하여 분류
- 이진 분류에서는 0~1 사이의 확률을 예측
- 예: 합격/불합격, 질병 O/X

## KNN
- 가까운 데이터들을 보고 예측
- 분류 / 회귀
- 예: 붓꽃 종류 분류

## Decision Tree
- 조건을 계속 나누면서 예측
- 분류 / 회귀

## Random Forest
- 여러 Decision Tree의 결과를 합쳐서 예측
- 분류 / 회귀

## SVM
- 데이터를 잘 나누는 경계를 찾아 예측
- 주로 분류
- 회귀(SVR)도 가능

## Gradient Boosting
- 이전 모델의 오차를 다음 모델이 계속 보완
- 분류 / 회귀

## XGBoost / LightGBM / CatBoost
- Gradient Boosting 계열
- 정형 데이터에서 많이 사용
- 분류 / 회귀

## K-Means
- 비슷한 데이터끼리 그룹화
- 군집화(Clustering)
- 정답(label)이 없는 비지도학습

## PCA
- 여러 특성을 적은 수의 특성으로 축소
- 차원 축소
- 데이터의 중요한 정보를 최대한 유지
